```python
from loguru import logger
import pandas as pd 


def merge_jcr_china(jcr_file: str, china_file: str, output_file: str):
    logger.info(f"Reading JCR data from {jcr_file}")
    jcr = pd.read_csv(jcr_file, encoding='utf-8')
    
    logger.info(f"Reading China data from {china_file}")
    chin = pd.read_csv(china_file, encoding='utf-8')
    
    logger.info("Merging JCR and China data")
    keyword: str = 'IF(2024)' if '2024' in jcr_file else 'IF(2025)'
    merged_df = (
        jcr
        .merge(right=chin, left_on='Journal', right_on='Journal', how='left')
        .replace({keyword: {'<0.1': 0}}) 
        .astype({keyword: 'float'})
        .sort_values(by=keyword, ascending=False)
        .query(f'`{keyword}` >= 1.0')
        [['Journal', keyword, 'Review', 'Open Access', '大类', '大类分区', '小类1', '小类1分区']]
        # .to_excel('./filtered_records/Summary_JCR2024-IF.xlsx', index=False)
    )
    
    logger.info(f"Saving merged data to {output_file}")
    merged_df.to_excel(output_file, index=False)
    logger.info("Merge completed successfully")

    return merged_df


# --- MAIN ---
# jcr_file: str = './raw_records/JCR2024-UTF8.csv'
jcr_file: str = './raw_records/JCR2025-UTF8.csv'

china_file: str = './raw_records/FQBJCR2025-UTF8.csv'

# output_file: str = './filtered_records/Merged_JCR_China_2024.xlsx'
output_file: str = './filtered_records/Merged_JCR_China_2025.xlsx'

merge_df = merge_jcr_china(jcr_file, china_file, output_file)
merge_df
```
